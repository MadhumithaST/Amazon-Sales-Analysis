```python
import pandas as pd
import numpy as np
df = pd.read_csv('amazon.csv')

#  Clean Price Columns (Remove ₹ and ,)
# We do this for both 'discounted_price' and 'actual_price'
def clean_currency(value):
    return float(str(value).replace('₹', '').replace(',', ''))
df['discounted_price'] = df['discounted_price'].apply(clean_currency)
df['actual_price'] = df['actual_price'].apply(clean_currency)

# Fix the 'rating' column
# One row has a "|" symbol which breaks calculations
df['rating'] = df['rating'].replace('|', np.nan)
df['rating'] = pd.to_numeric(df['rating'], errors='coerce')
df['rating'] = df['rating'].fillna(df['rating'].median())

#  Clean 'rating_count'
df['rating_count'] = df['rating_count'].str.replace(',', '')
df['rating_count'] = pd.to_numeric(df['rating_count'], errors='coerce')

#  Simplify Categories for Power BI
# The 'category' column is too long (e.g., Computers|Accessories|Cables)
# We only want the first part (e.g., Computers)
df['main_category'] = df['category'].str.split('|').str[0]

df.to_csv('amazon_cleaned.csv', index=False)
print("Success! Cleaned file ready for Power BI.")
```
